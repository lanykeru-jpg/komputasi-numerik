```python
def eliminasi_gauss(A, b):
    n = len(b)

    # Menggabungkan matriks A dan vektor b menjadi augmented matrix [A | b]
    for i in range(n):
        A[i].append(b[i])

    # Proses Eliminasi Maju (Forward Elimination)
    for i in range(n):
        # Partial Pivoting untuk menghindari pembagian dengan nol
        max_el = abs(A[i][i])
        max_row = i
        for k in range(i + 1, n):
            if abs(A[k][i]) > max_el:
                max_el = abs(A[k][i])
                max_row = k

        # Tukar baris jika diperlukan
        A[i], A[max_row] = A[max_row], A[i]

        # Eliminasi elemen di bawah diagonal utama
        for k in range(i + 1, n):
            c = -A[k][i] / A[i][i]
            for j in range(i, n + 1):
                if i == j:
                    A[k][j] = 0
                else:
                    A[k][j] += c * A[i][j]

    # Proses Substitusi Balik (Back Substitution)
    x = [0 for _ in range(n)]
    for i in range(n - 1, -1, -1):
        x[i] = A[i][n]
        for j in range(i + 1, n):
            x[i] -= A[i][j] * x[j]
        x[i] = x[i] / A[i][i]

    return x

# Contoh Penggunaan SPLTV:
# 2x + y + z = 4
# x + 3y + 2z = 6
# x + y + 3z = 5
A = [
    [2, 1, 1],
    [1, 3, 2],
    [1, 1, 3]
]
b = [4, 6, 5]

hasil = eliminasi_gauss(A, b)
print("Penyelesaian Sistem Persamaan Linear Tiga Variabel:")
print(f"x = {hasil[0]:.4f}")
print(f"y = {hasil[1]:.4f}")
print(f"z = {hasil[2]:.4f}")
```
